```python
# Import required libraries
from langdetect import detect, DetectorFactory

# Ensure consistent results
DetectorFactory.seed = 0

# Language code and language name dictionary
LANGUAGES = {
    "en": "English",
    "ta": "Tamil",
    "te": "Telugu",
    "hi": "Hindi",
    "ml": "Malayalam",
    "kn": "Kannada"
}


# Language detection function
def detect_language(text):

    try:
        # Detect language code
        language_code = detect(text)

        # Convert code into complete language name
        language_name = LANGUAGES.get(
            language_code,
            "Unknown Language"
        )

        return language_code, language_name

    except Exception:
        return None, "Language could not be detected"


# Main program
print("===== MULTILINGUAL LANGUAGE DETECTION =====")

# Get input from user
text = input("Enter your text: ")

# Check empty input
if not text.strip():

    print("Please enter some text.")

else:

    # Call language detection function
    code, name = detect_language(text)

    # Display results
    print("\n----- DETECTION RESULTS -----")

    print("Input Text:", text)

    print("Language Code:", code)

    print("Detected Language:", name)
```
